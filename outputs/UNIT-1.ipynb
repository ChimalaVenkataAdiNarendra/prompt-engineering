```python
# 1. Environment & Connectivity: Install required packages (e.g., transformers, openai); securely configure the API key; run a simple "Hello, world" prompt to verify model access.

!pip install -q openai pandas

from google.colab import userdata
from openai import OpenAI

MAX_CALLS = 40
call_count = 0

client = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=userdata.get("GROQ_API_KEYS"),
)
MODEL = "llama-3.3-70b-versatile"


def ask_llm(prompt, system=None, temperature=0.3):
    """Send a prompt to Groq. Raises if daily call cap exceeded."""
    global call_count
    call_count += 1
    if call_count > MAX_CALLS:
        raise RuntimeError(f"Lab limit reached ({MAX_CALLS} calls). Stop and save your work.")
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})
    response = client.chat.completions.create(
        model=MODEL,
        messages=messages,
        temperature=temperature,
    )
    return response.choices[0].message.content

print("Setup complete. Model:", MODEL)

# Your first API call (run with instructor) and
prompt = "Say Hello to IV B.Tech CSE students in exactly one sentence."

print("USER PROMPT:")
print(prompt)
print("\nMODEL RESPONSE:")
response = ask_llm(prompt)
print(response)

# ---Simple "Hello, world" prompt to verify model access.  ---
response= (ask_llm("Hello World"))
print(response)

```

```python
# 2. Baseline vs. Enhanced Prompts: Execute a naïve prompt ("Write a one-paragraph bio of Ada Lovelace.") and an enhanced prompt that adds role framing, specificity, and explicit format instructions; compare both outputs for relevance, completeness, and style.

#  Baseline prompt vs enhanced prompt (run with instructor)
naive = "Write a one-paragraph bio of Ada Lovelace."

enhanced = """You are a computer science historian. Write a one-paragraph biography of Ada Lovelace in 120–150 words
- her full name and lifespan,
- her collaboration with Charles Babbage,
- why she is considered the first computer programmer,
- her lasting impact on modern computing.
Use a formal academic tone and end with one sentence summarizing her legacy."""

print("=== NAIVE PROMPT ===")
print(naive)
print("\nOUTPUT:")
print(ask_llm(naive))

print("\n" + "=" * 50)
print("=== ENHANCED PROMPT ===")
print(enhanced)
print("\nOUTPUT:")
print(ask_llm(enhanced))

```

```python
# 3. Iterative Refinement on a Simple Task: Summarize the plot of the Shakespearean play Romeo and Juliet in two sentences through three rounds of prompt tweaking:

# a. Minimal instruction.

# b. Addition of length and style constraints

# c. Specification of key content elements (setting and theme)

# Document how each iteration changes and improves the result.

prompts = {
    "v1": "Provide a concise summary of Romeo and Juliet.",

    "v2": "Summarize  Romeo and Juliet in exactly two sentences using simple and clear English.",

    "v3": "Summarize  Romeo and Juliet in exactly two sentences using simple English. Mention that the story is set in Verona, highlight the theme of forbidden love, and briefly describe the conflict and tragic ending."
}

for version, prompt in prompts.items():
    print(f"\n{version.upper()} Prompt:")
    print(prompt)

    print("\nResponse:")
    print(ask_llm(prompt))


```
